```python
class Vehicle:
    """A single parent class shared by multiple child classes."""
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        
    def display_info(self):
        return f"This is a {self.brand} {self.model}."
```

```python
class Car(Vehicle):
    """Inherits from Vehicle and adds specific functionality for cars."""
    def __init__(self, brand, model, doors):
        # Call the parent class constructor
        super().__init__(brand, model)
        self.doors = doors
        
    def drive(self):
        return f"The {self.model} car is driving smoothly on 4 wheels."
```

```python
class Motorcycle(Vehicle):
    """Inherits from the SAME Vehicle class, but has different features."""
    def __init__(self, brand, model, has_sidecar=False):
        # Call the parent class constructor
        super().__init__(brand, model)
        self.has_sidecar = has_sidecar
        
    def ride(self):
        return f"The {self.model} motorcycle is weaving through traffic."
```

```python
my_car = Car(brand="Tesla", model="Model 3", doors=4)
print("--- Car Object ---")
print(my_car.display_info())  # Inherited method
print(my_car.drive())         # Child-specific method
print(f"Doors: {my_car.doors}\n")
```

```python
my_bike = Motorcycle(brand="Ducati", model="Panigale")
print("--- Motorcycle Object ---")
print(my_bike.display_info())  # Inherited method (from same parent)
print(my_bike.ride())         # Child-specific method
print(f"Has sidecar: {my_bike.has_sidecar}")
```

```python
class OTTPsubscription:
    def __init__(self, subscription_id, plan, total_payment):
        self.id = subscription_id
        self.plan_package = plan
        self.total_payment = total_payment

    def subscription(self):
        print (f"Subscriber with {self.id} is subscribed to the {self.plan_package}")

    def unsubscribe(self):
        print (f"Subscriber with {self.id} unsubscribed to the {self.plan_package}")
```

```python
class PermiumSubscription(OTTPsubscription):
    def __init__(self, subscription_id, plan, total_payment, screens):
        super().__init__(subscription_id, plan, total_payment)
        self.max_screens = screens
    def set_max_screen(self):
        print(f"maximum screens for this subscription is {self.max_screens}")
```

```python
obj = PermiumSubscription(122,"pro",8000,7)
```

```python
obj.set_max_screen()
```
