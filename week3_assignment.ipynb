```python
# Task 1: inventory dictionary (item name -> stock count)
inventory = {"Widget": 10, "Gadget": 5, "Sensor": 0, "Cable": 15, "Battery": 8}

print("Starting inventory:", inventory)
```

```python
# Task 1: order queue (list of lists: [item, quantity])
orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 15],
    ["Router", 4],
    ["Battery", 5],
    ["Widget", 9],
    ["Cable", 1],
]

print("Number of orders in queue:", len(orders))
print(orders)
```

```python
# Variables used to track results
fully_fulfilled = 0   # counts orders filled completely
unfulfilled = []      # stores [item, quantity still owed]
```

```python
# Task 2: process each order
print("=== PROCESSING ORDERS ===")

for order in orders:
    item = order[0]
    qty = order[1]
    stock = inventory.get(item)   # returns None if the item doesn't exist

    if stock is None:
        # Invalid item
        print(f"ALERT: '{item}' does not exist in inventory. Order of {qty} rejected.")
        unfulfilled.append([item, qty])

    elif stock >= qty:
        # Full fulfillment
        inventory[item] = stock - qty
        fully_fulfilled += 1
        print(f"SUCCESS: {qty} x {item} shipped. Remaining stock: {inventory[item]}")

    elif stock > 0:
        # Partial fulfillment
        remainder = qty - stock
        print(f"PARTIAL: Only {stock} of {qty} {item} available. Shipped {stock}, short by {remainder}.")
        inventory[item] = 0
        unfulfilled.append([item, remainder])

    else:
        # Out of stock
        print(f"OUT OF STOCK: {item} is sold out. Cannot ship {qty}.")
        unfulfilled.append([item, qty])
```

```python
# Task 3: end-of-batch summary
print("=== END OF BATCH SUMMARY ===")
print("Final inventory:", inventory)
print("Fully fulfilled orders:", fully_fulfilled)
print("Items not fully supplied:")

for entry in unfulfilled:
    print(f"  - {entry[0]}: {entry[1]} unit(s) short")
```
