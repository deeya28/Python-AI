```python
# Task 1a: remove non-positive amounts and amounts below min_threshold
def clean_transactions(amounts, min_threshold=10.0):
  result = filter(lambda x: x > 0 and x >= min_threshold, amounts)
  return list(result)
```

```python
raw = [250.0, -40.0, 5.5, 0, 10.0, 99.99, 3.0, 1200.0]
cleaned = clean_transactions(raw)
cleaned
```

```python
# Task 1b: deduct a percentage fee from every amount
def apply_service_fee(amounts, fee_rate=0.02):
  result = map(lambda x: round(x * (1 - fee_rate), 2), amounts)
  return list(result)
```

```python
processed = apply_service_fee(cleaned)
processed
```

```python
print(clean_transactions(raw, min_threshold=100))
print(apply_service_fee([200], fee_rate=0.1))
```

```python
def calculate_nested_total(data):
  # base case 1: empty list
  if data == []:
    return 0.0
  # base case 2: single number
  if isinstance(data, (int, float)):
    return data
  # recursive step: split into head and tail
  head = data[0]
  tail = data[1:]
  return calculate_nested_total(head) + calculate_nested_total(tail)
```

```python
print(calculate_nested_total([100, [50, 25], [[10, 5], 40]]))
print(calculate_nested_total([[[[500]]], 20.5, [[], [30, [[]]]]]))
print(calculate_nested_total([]))
print(calculate_nested_total(75.25))
```
