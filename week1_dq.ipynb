```python
import time

def expensive_check(data):
    print("Running expensive_check...")
    time.sleep(1)  # simulate a slow operation, like a DB call
    return data > 5

def cheap_check(flag):
    print("Running cheap_check...")
    return flag is True

def process():
    print("Processing...")

data = 10
flag = True

# Less efficient: expensive_check may run needlessly
if expensive_check(data) and cheap_check(flag):
    process()

print("---")

# More efficient: cheap_check runs first, so expensive_check only runs if needed
if cheap_check(flag) and expensive_check(data):
    process()
```

```python
s = "hello"

try:
    s[0] = "H"
except TypeError as e:
    print("Error:", e)      # 'str' object does not support item assignment

t = s.upper()                # makes a NEW string
print(s, id(s))              # hello  (unchanged)
print(t, id(t))              # HELLO  (different object, different id)
```

```python
import timeit

words = ["word"] * 100_000

def concat():
    result = ""
    for w in words:
        result += w
    return result

def join():
    return "".join(words)

print("+= loop:", timeit.timeit(concat, number=5) / 5)
print("join   :", timeit.timeit(join, number=5) / 5)
```
