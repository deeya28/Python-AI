```python
# Caesar Log Decryptor - decrypts raw_logs.txt (left shift of 3)
# and flags any line containing "BREACH".

infile = open("raw_logs.txt", "r")
master = open("decrypted_master.txt", "w")
alerts = open("security_alerts.txt", "w")

for line in infile:
    decrypted = ""

    for ch in line:
        if ch == " " or ch == "\n":
            decrypted += ch                      # preserve spaces/newlines
        elif "A" <= ch <= "Z":
            code = ord(ch) - 3
            if code < ord("A"):                  # wrap-around: A -> X, B -> Y, C -> Z
                code += 26
            decrypted += chr(code)
        elif "a" <= ch <= "z":
            code = ord(ch) - 3
            if code < ord("a"):
                code += 26
            decrypted += chr(code)
        else:
            decrypted += ch                      # digits/punctuation unchanged

    decrypted = decrypted.strip()

    master.write(decrypted + "\n")

    if "BREACH" in decrypted.upper():
        alerts.write(decrypted + "\n")

infile.close()
master.close()
alerts.close()

print("Done: decrypted_master.txt and security_alerts.txt created.")
```
