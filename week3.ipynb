```python
def update_list():
    my_list = []

    for i in range(5):
        value = input("Enter a number: ")

        if "." in value:
            # Float number
            float_value = float(value)

            if my_list:
                my_list.pop()
                print("Float entered. Last element removed.")
            else:
                print("Float entered, but the list is empty.")

        else:
            # Integer number
            int_value = int(value)
            my_list.append(int_value)
            print("Integer added to the list.")

        print("Updated list:", my_list)


# Function call
update_list()
```

```python
def student_result():
    students = {}

    # Take input for 5 students
    for i in range(5):
        name = input("Enter student name: ")
        marks = float(input("Enter marks: "))

        # Assign grade
        if marks >= 80:
            grade = "A"
        elif marks >= 60:
            grade = "B"
        elif marks >= 40:
            grade = "C"
        else:
            grade = "F"

        # Store data in dictionary
        students[name] = {"marks": marks, "grade": grade}

    # Display student results
    print("\n--- Student Results ---")

    for name, data in students.items():
        print("Name:", name)
        print("Marks:", data["marks"])
        print("Grade:", data["grade"])
        print()

    # Calculate highest, lowest, and average
    marks_list = [data["marks"] for data in students.values()]

    highest = max(marks_list)
    lowest = min(marks_list)
    average = sum(marks_list) / len(marks_list)

    print("Highest Marks:", highest)
    print("Lowest Marks:", lowest)
    print("Average Marks:", average)


# Function call
student_result()
```

```python
#factorial number
# 5! = 5*4*3*2*1 
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
factorial(5)
```

```python
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_sequence = [0, 1]
        for i in range(2, n):
            next_number = fib_sequence[i - 1] + fib_sequence[i - 2]
            fib_sequence.append(next_number)
        return fib_sequence
```

```python
fibonacci(10)
```

```python
def myWrapper(func):
    def myInnerFunc():
        print("Inside wrapper.")
        func()
    return myInnerFunc
```

```python
@myWrapper
def myFunc():
    print("Hello, World")
```

```python
myFunc()
```

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper
@logger
def greet(name, age):
    print(f"Hello {name}, Age:{age}")

greet("Diya", 21)
```

```python
def count_up(n):
    i=1
    while i<=n:
        yield i
        i+=1
```
