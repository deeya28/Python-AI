```python
import random
f = open("integer.txt", 'w')
for count in range(500):
    number = random.randint(1,500)
    f.write(str(number)+"\n")
f.close()
```

```python
f = open("integer.txt", 'r')
sum = 0

for line in f:
    line = line.strip()
    number = int(line)
    sum +=number
print("The sum is", sum)
f.close()
```

```python
f = open("integer.txt", 'r')
sum = 0

for line in f:
    wordlist = line.strip().split()
    number = int(wordlist[0])
    sum +=number
print("Sum is", sum)
f.close()    
```
